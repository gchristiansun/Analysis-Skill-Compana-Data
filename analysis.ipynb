```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

from utils.data_loader import load_data
from utils.visualization import (
    plot_top_roles,
    plot_problem_distribution
)
```

```python
def load_data(path):
    df = pd.read_csv(path)

    if "Unnamed: 0" in df.columns:
        df = df.drop(columns=["Unnamed: 0"])

    return df
```

```python
def plot_top_roles(df, save_path):

    role_count = (
        df["target_role"]
        .value_counts()
        .head(10)
    )

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=role_count.values,
        y=role_count.index
    )

    plt.title("Top 10 Target Roles")
    plt.xlabel("Jumlah")
    plt.ylabel("Role")

    plt.tight_layout()
    plt.savefig(save_path)
    plt.close()


def plot_problem_distribution(df, save_path):

    problem_count = (
        df["problem_category"]
        .value_counts()
    )

    plt.figure(figsize=(8, 8))

    plt.pie(
        problem_count.values,
        labels=problem_count.index,
        autopct="%1.1f%%"
    )

    plt.title("Problem Category Distribution")

    plt.savefig(save_path)
    plt.close()
```
